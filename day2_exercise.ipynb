```python
# Declare two variables, `x` and `y`, with values 20 and 4, respectively. 
# Calculate their sum, difference, product, and quotient.
x = 20
y = 4
sum = x + y
product = x * y
quotient = x/y
print(f"Sum: {sum}, Product: {product}, Quotient: {quotient}")
```

```python
# Write a Python program that asks the user for their age and checks if they are eligible to vote (age >= 18).
user_age = int(input("Hello, what is your age: "))
if user_age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote!")
```

```python
# Create a variable `pi` with the value 3.14159. Use it to calculate and print the area of a circle with radius 7.
pi = 3.14159
radius = 7
area_of_circle = pi * radius ** 2
print(area_of_circle)
```

```python
# Write a program that takes two numbers from the user and 
# prints whether the first number is greater than, less than, or equal to the second number.
x = int(input("Enter the first number: "))
y = int(input("Enter the second number: "))

if x > y:
    print(f"Your {x} is greater than {y}")
elif x < y:
    print(f"Your {x} is less than {y}")
else:
    print(f"{x} and {y} are equal")
```

```python
# Declare a Boolean variable called `is_raining` and set it to `True` or `False`. 
# Use it to print "Take an umbrella" if it is raining, otherwise print "Enjoy the sunshine".
is_raining = False
if is_raining == True:
    print("Take an umbrella it is raining")
else:
    print("Enjoy the sunshine")
```

```python
# Write a Python program that takes two numbers from the user and prints the result of their division. 
# Make sure to handle the case when the second number is zero.
a = int(input("Enter a number: "))
b = int(input("Enter your second number: "))

if b == 0:
    print("The second number cannot be 0. Math Error")
else:
    division = a / b
    print(f"The quotient of those two numbers is: {division}")
```

```python
# Write a Python program that converts Celsius to Fahrenheit. Formula: `F = (C * 9/5) + 32`.
temp_celsius = int(input("Enter the temperature in Celsius: "))
converter = (temp_celsius * 9/5) + 32
print(f"Your temperature in Farenheit is: {converter}")
```

```python
# Create a Python program that takes a number from the user and checks whether it is even or odd.
number = int(input("Enter a number: "))
if number % 2 == 0:
    print(f"Your number {number} is an Even Number")
else:
    print(f"Your number {number} is an odd number")
```

```python
# Write a program that takes a string from the user and 
# prints whether it is a palindrome (the same forwards and backwards).
message = input("Enter a word. We will check if is a palindrome: ")
if message == message[::-1]:
    print(f"The word {message} is a palindrome")
else:
    print(f"The word {message} is not a palindrome")
```
