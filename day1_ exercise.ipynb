3. Basic Input  
Write a program that asks for the user's age and prints a message saying "You are X years old" (replace X with the input).

```python
try:
    user_age = int(input("What is your age: "))
    print(f"You are {user_age} years old.")
except ValueError:
    print("Please enter a valid number")
```

4. Simple Arithmetic Operations  
Write a program that takes two numbers as input and prints their sum, difference, multiplication, and division.

```python
num_a = int(input("Enter a number: "))
num_b = int(input("Enter a number: "))
sum = (num_a + num_b)
difference = (num_a - num_b)
multiplication = (num_a * num_b)
division = (num_a / num_b)
print(f"Your arithmetic operations for the two number are Sum: {sum}, Difference: {difference}, Multiplication: {multiplication}, Division: {division}")
```
